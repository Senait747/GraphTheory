````python
# Step 4: Build public URLs for audio files
def build_public_url(row):
    """Build public URL for audio file in Supabase storage."""
    base_url = f"{url}/storage/v1/object/public/{BUCKET_NAME}"
    object_name = row.get('object_name', '')
    if object_name:
        return f"{base_url}/{object_name}"
    return base_url

print("\n Building audio URLs...")
df["audio_url"] = df.apply(build_public_url, axis=1)
print(" Added audio_url column")
print("\nExample URL:")
print(df['audio_url'].iloc[0])
````

````python
# Step 5: Define feature extraction function
def extract_mfcc_features(file_url, n_mfcc=20, duration=5.0):
    """
    Download audio file and extract MFCC features.

    Args:
        file_url: URL to audio file
        n_mfcc: Number of MFCC coefficients to extract
        duration: Duration of audio to process (seconds)

    Returns:
        numpy array of mean MFCC features
    """
    try:
        # Download audio with timeout
        r = requests.get(file_url, timeout=30)
        r.raise_for_status()

        # Load audio from memory (no temp files!)
        y, sr = librosa.load(BytesIO(r.content), duration=duration)

        # Validate audio
        if len(y) == 0:
            return np.zeros(n_mfcc)

        # Extract MFCC features
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
        mfccs_mean = np.mean(mfccs.T, axis=0)

        return mfccs_mean

    except Exception as e:
        print(f"❌ Error: {str(e)[:50]}")
        return np.zeros(n_mfcc)
````

````python
# Step 7: Clean data - remove failed extractions
print("\n Cleaning data...")
valid_indices = ~np.all(X == 0, axis=1)
X_clean = X[valid_indices]
df_clean = df[valid_indices].reset_index(drop=True)

print(f" Valid samples: {len(X_clean)}/{len(X)}")
print(f"   Removed {len(X) - len(X_clean)} failed extractions")
````

## Step 6: Extract Audio Features (MFCC)

### What are MFCCs?
**MFCC = Mel-Frequency Cepstral Coefficients**

These are numbers that describe what an audio recording sounds like. Instead of storing the entire audio file, we extract 20 numbers that capture the important characteristics of the sound.

**Think of it like:** Describing a person with "height, weight, hair color, eye color" instead of showing their photo.

### The Function

```python
def extract_mfcc_features(file_url, n_mfcc=20, duration=5.0):
    try:
        # Download audio
        r = requests.get(file_url, timeout=30)
        r.raise_for_status()
        
        # Load audio from memory
        y, sr = librosa.load(BytesIO(r.content), duration=duration)
        
        # Validate audio
        if len(y) == 0:
            return np.zeros(20)
        
        # Extract MFCC features
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)
        mfccs_mean = np.mean(mfccs.T, axis=0)
        
        return mfccs_mean
        
    except Exception as e:
        return np.zeros(20)
```

**Line by line:**

1. `requests.get(file_url, timeout=30)` - Download the audio file from the URL
2. `r.raise_for_status()` - Check if download was successful
3. `librosa.load(BytesIO(r.content), duration=5.0)` - Convert downloaded bytes into audio data (only first 5 seconds)
   - `y` = audio signal (array of numbers representing sound wave)
   - `sr` = sample rate (how many measurements per second)
4. `if len(y) == 0:` - Check if audio is empty
5. `librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)` - Extract 20 MFCC coefficients from the audio
6. `np.mean(mfccs.T, axis=0)` - Take the average of MFCCs across time (gives 20 numbers)
7. `except Exception as e:` - If anything fails, return 20 zeros

**What you get:** 20 numbers that describe the audio, like `[12.5, -3.2, 8.1, ...]`

### Input:
- Database with 1887 frog audio files
- Each file has a species label

### Process:
1. Download file information from database
2. Download each audio file
3. Extract 20 numbers (MFCCs) that describe each sound
4. Remove any files that failed
5. Convert species names to numbers
6. Split into training (80%) and testing (20%) sets
7. Save everything

### Output:
- **X_train**: 1509 audio files × 20 features (numbers)
- **X_test**: 378 audio files × 20 features
- **y_train**: 1509 labels (which species each audio is)
- **y_test**: 378 labels

### Next Steps:
Now you can train a machine learning model:
```python
# Load the data
X_train = np.load('X_train.npy')
y_train = np.load('y_train.npy')

# Train a model (example with Random Forest)
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
model.fit(X_train, y_train)

# Test accuracy
X_test = np.load('X_test.npy')
y_test = np.load('y_test.npy')
accuracy = model.score(X_test, y_test)
print(f"Accuracy: {accuracy * 100}%")
```

## Key Concepts for Beginners

### 1. Features vs Labels
- **Features (X)**: Input data - the 20 MFCC numbers describing audio
- **Labels (y)**: Output data - which species the audio belongs to

### 2. Why Machine Learning?
The model learns patterns in the features that distinguish different frog species. Like learning that "Green Frogs have higher pitch" or "Bullfrogs have deeper sounds."

### 3. Train vs Test
- **Train**: Show the model examples so it can learn
- **Test**: Give it new examples to see if it really understood

Think of it like: studying with practice problems (train) then taking an exam with new problems (test).

## Common Beginner Questions

**Q: Why 20 MFCCs?**
A: It's a standard number that captures enough audio information without being too complex. You could use 13 or 40, but 20 is a good balance.

**Q: What if some files fail?**
A: That's normal! Network issues happen. The code handles it by returning zeros, and we remove those in the cleaning step.

**Q: Can I use this for other sounds?**
A: Yes! Just change the database and labels. Works for bird songs, music genres, speech, etc.

**Q: How long does this take?**
A: For 1887 files, expect 1-2 hours depending on your internet speed. The progress bar shows you how much is left.

**Q: What's next after this?**
A: Train different models (Random Forest, SVM, Neural Networks) and compare their accuracy!
